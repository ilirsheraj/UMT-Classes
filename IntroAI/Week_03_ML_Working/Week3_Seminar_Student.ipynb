## Part 4 — Your first model, built by hand

In the lecture: **a model is a function**. Inputs go in, a prediction comes out.

In Python, `def` creates a function. Here is a very simple model: *price = size × a price per m²*.

```
def name(input):
    return output
```

**Fill in the blank:** start with a guess of **1000** euros per m².

````python
# rate in euros per square metre — our model's only setting
rate = 2000

def predict_price(size):
    return size * rate

predict_price(95)
````

The real price of that flat is **210,000**. Our model is quite wrong for this flat. But is it wrong *on average*?

### How wrong is the model? Average error

For every flat: **error = |prediction − real price|** (the `abs()` function removes the minus sign).

Then we average the errors — exactly the "average error" from the lecture.

A `for` loop repeats the indented lines once **for each** flat in the list.

**Fill in the blank:** the error compares the guess with the **real** price of the flat.

````python
total_error = 0

for f in flats:
    guess = predict_price(f["size"])
    error = abs(guess - f["price"])
    total_error = total_error + error

average_error = total_error / len(flats)
print("Average error: €", round(average_error))
````

## Part 6 — The black box: a real ML model

Now the same data, the professional way. You **do not** need to understand every line yet. Match each step to the **ML workflow** from the lecture.

**Step: load the data.** `pandas` turns our list of dictionaries into a table (a *DataFrame*).

````python
import pandas as pd

data = pd.DataFrame(flats)
# show the first 5 rows
data.head()
````

**Steps: choose features and label, then split train / test.** 75% of flats for training, 25% hidden for testing.

````python
from sklearn.model_selection import train_test_split

# features
X = data[["size", "rooms"]]
# label
y = data["price"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
print("Training flats:", len(X_train), "  Test flats:", len(X_test))
````

**Steps: train, then predict.** `fit` = training, `predict` = prediction.

````python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
# training
model.fit(X_train, y_train)
# prediction on flats it has never seen
predictions = model.predict(X_test)
````

**Step: evaluate.** Average error on the **test** flats — for the ML model and for **your** hand-made model.

Set `rate` to the best value you found in Part 4.

````python
# my best rate from Part 4
rate = 1450

ml_error   = abs(predictions - y_test).mean()
hand_error = abs(X_test["size"] * rate - y_test).mean()

print("ML model average error:     €", round(ml_error))
print("Your model average error:   €", round(hand_error))
````
